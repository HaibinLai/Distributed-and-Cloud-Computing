```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
from pyspark.sql import DataFrameReader
from pyspark.sql import functions as F, Window

# reading data from a CSV
data_reader = DataFrameReader(spark)
df = data_reader.csv("Pokemon.csv", header=True, )
df.show(3)
```

```python
# count number of fire pokemon
df.filter(F.col("Type 1") == "Fire").count()
```

```python
# Average total stats of all pokemon in each category
(
df.groupBy(
    F.col("Type 1"))
    .agg(
        F.avg(F.col("Total"))
        .alias("Average Total Stats of All Pokemon in Type")
    )
    .show()
)
```
